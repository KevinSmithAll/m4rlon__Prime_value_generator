# Генератор простых чисел из заданного промежутка

```python
def get_list_light_values(left: int, right: int) -> list:
    final_list = []
    for i in range(left, right + 1, 1):
        if i == 0 or i == 1 or i == 2 or i == 3:
            final_list.append(i)
            continue
        for j in range(2, i):
            if i % j == 0:
                break
            if j == i - 1:
                final_list.append(i)
    return final_list
```

```python
get_list_light_values(1, 20)
```
